# Hello World Experimenty

Základní experimenty s print funkcí a escape sekvencemi v Pythonu.

## Základní Hello World

```python
# Hello world
print("Hello World!")
```

## Různé způsoby práce s uvozovkami

```python
print("Hello \"World!\"")
print("Hello 'World!'")
print('Hello "World!"')
```

## Escape sekvence - nový řádek a tabulátor

```python
print("First line\nSecond line")
print("First line\tSecond line\a")
```

## Zvukový signál (BELL) - různé způsoby

Zvukový signál není všude podporován, ale můžeme ho vyzkoušet různými způsoby:

```python
# Escape sekvence \a
print("\a")
```

```python
# Pomocí ASCII znaku (hexadecimální notace)
print("\x07")
```

```python
# Unicode varianta
print("\u0007")
```

```python
# Emoticon zvonku (pojmenovaný unicode znak)
print("\N{BELL}")
```
