# Formátování řetězců v Pythonu

Přehled všech možností formátování a prefixů řetězců.

## 1. f-strings (formatted strings) - Python 3.6+

Nejmodernější a nejčitelnější způsob formátování.

```python
# Základní použití
name = "World"
print(f"Hello {name}!")

# Výrazy uvnitř {}
print(f"2 + 2 = {2 + 2}")
print(f"Uppercase: {name.upper()}")
```

```python
# Formátování čísel
pi = 3.14159265359
print(f"Pi na 2 des. místa: {pi:.2f}")
print(f"Pi na 4 des. místa: {pi:.4f}")

# Zarovnání a šířka
print(f"{'Left':<10}|")
print(f"{'Center':^10}|")
print(f"{'Right':>10}|")
```

```python
# Čísla v různých soustavách
num = 42
print(f"Decimal: {num:d}")
print(f"Binary: {num:b}")
print(f"Octal: {num:o}")
print(f"Hex: {num:x}")
print(f"Hex upper: {num:X}")
```

## 2. r-strings (raw strings)

Ignorují escape sekvence - užitečné pro cesty a regulární výrazy.

```python
# Bez raw string
print("C:\\new\\path")  # musíme escapovat backslash

# S raw string
print(r"C:\new\path")  # \ se neinterpretuje

# Escape sekvence zůstávají doslovné
print(r"\n zůstane jako \n, ne nový řádek")
```

```python
# Praktické použití - regex patterns
import re

pattern = r"\d+\.\d+"  # hledání desetinných čísel
text = "Pi je přibližně 3.14"
match = re.search(pattern, text)
if match:
    print(f"Nalezeno: {match.group()}")
```

## 3. b-strings (bytes)

Bajtové řetězce pro práci s binárními daty.

```python
# Bytes literal
byte_string = b"Hello"
print(byte_string)
print(type(byte_string))

# Převod string -> bytes
text = "Ahoj"
bytes_text = text.encode('utf-8')
print(bytes_text)

# Převod bytes -> string
decoded = bytes_text.decode('utf-8')
print(decoded)
```

```python
# Práce s jednotlivými bajty
data = b"ABC"
for byte in data:
    print(f"Byte: {byte}, Char: {chr(byte)}")
```

## 4. u-strings (unicode)

V Python 3 už není potřeba - všechny stringy jsou unicode.

```python
# V Python 3 jsou tyto ekvivalentní
unicode_str = u"Příliš žluťoučký kůň"
normal_str = "Příliš žluťoučký kůň"

print(unicode_str == normal_str)
print(type(unicode_str))
print(type(normal_str))
```

## 5. Triple quotes (""" nebo ''')

Pro víceřádkové řetězce.

```python
# Víceřádkový text
text = """První řádek
Druhý řádek
Třetí řádek"""
print(text)

# Často používáno pro docstrings
def my_function():
    """
    Toto je docstring.
    Může být na více řádků.
    """
    pass

print(my_function.__doc__)
```

## 6. Kombinace prefixů

Některé prefixy lze kombinovat.

```python
# fr nebo rf - raw f-string
path = "documents"
print(fr"C:\Users\{path}\file.txt")
print(rf"C:\Users\{path}\file.txt")  # stejné jako fr

# rb nebo br - raw bytes
print(rb"C:\path\to\file")
```

## 7. Starší způsoby formátování

Stále platné, ale f-stringy jsou preferované.

```python
# .format() metoda - Python 2.6+
name = "World"
age = 5

print("{} is {} years old".format(name, age))
print("{0} is {1} years old".format(name, age))
print("{name} is {age} years old".format(name=name, age=age))
```

```python
# % operátor - starý C-style (Python 2)
print("Hello %s" % name)
print("%s is %d years old" % (name, age))
print("Pi: %.2f" % 3.14159)
```

## 8. Praktické příklady kombinací

```python
# Formátovaná tabulka
products = [
    ("Jablka", 5, 12.50),
    ("Hrušky", 3, 15.00),
    ("Banány", 8, 8.25)
]

print(f"{'Produkt':<10} {'Počet':>6} {'Cena':>8}")
print("-" * 26)
for product, count, price in products:
    print(f"{product:<10} {count:>6} {price:>8.2f}")
```

```python
# Debug výpis s = (Python 3.8+)
x = 10
y = 20
print(f"{x=}")
print(f"{y=}")
print(f"{x + y=}")
```

```python
# Datum a čas formátování
from datetime import datetime

now = datetime.now()
print(f"Aktuální čas: {now:%Y-%m-%d %H:%M:%S}")
print(f"Datum: {now:%d.%m.%Y}")
print(f"Čas: {now:%H:%M}")
```

## Shrnutí

| Prefix | Použití | Příklad |
|--------|---------|----------|
| `f""` | Formátované stringy | `f"Hello {name}"` |
| `r""` | Raw stringy (ignorují escape) | `r"C:\path"` |
| `b""` | Bytes | `b"Hello"` |
| `u""` | Unicode (Python 3 default) | `u"Text"` |
| `""""""` | Víceřádkové | `"""line1\nline2"""` |
| `fr""` | Raw f-string | `fr"C:\{folder}"` |
| `rb""` | Raw bytes | `rb"C:\data"` |

**Doporučení:** Pro běžné formátování používej **f-stringy** - jsou nejčitelnější a nejrychlejší!
