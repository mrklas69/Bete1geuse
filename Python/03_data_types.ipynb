# Datové typy v Pythonu

Přehled základních datových typů a jejich vlastností.

## 1. Základní datové typy (Primitives)

Python má několik základních datových typů.

### Integer (int) - celá čísla

```python
# Integer (celá čísla)
x = 42
y = -17
big_number = 1_000_000_000  # podtržítka pro čitelnost

print(f"x = {x}, typ: {type(x)}")
print(f"y = {y}, typ: {type(y)}")
print(f"Velké číslo: {big_number}")
```

```python
# Různé soustavy
binary = 0b1010  # binární (2)
octal = 0o12     # osmičková (8)
hexa = 0xFF      # hexadecimální (16)

print(f"Binary 0b1010 = {binary}")
print(f"Octal 0o12 = {octal}")
print(f"Hex 0xFF = {hexa}")
```

### Float - desetinná čísla

```python
# Float (desetinná čísla)
pi = 3.14159
temperature = -15.5
scientific = 1.5e-4  # vědecká notace (1.5 * 10^-4)

print(f"Pi: {pi}, typ: {type(pi)}")
print(f"Teplota: {temperature}°C")
print(f"Vědecká notace: {scientific}")
```

```python
# Operace s float
a = 10.5
b = 2.5

print(f"Součet: {a + b}")
print(f"Rozdíl: {a - b}")
print(f"Součin: {a * b}")
print(f"Podíl: {a / b}")
print(f"Mocnina: {a ** 2}")
```

### String (str) - textové řetězce

```python
# String (textové řetězce)
text = "Hello, World!"
multiline = """Tento text
může být
na více řádků"""

print(f"Text: {text}, typ: {type(text)}")
print(f"Délka: {len(text)}")
print(f"\nVíceřádkový text:\n{multiline}")
```

```python
# Operace se stringy
first = "Python"
second = "Programming"

print(f"Spojení: {first + ' ' + second}")
print(f"Opakování: {first * 3}")
print(f"Velká písmena: {first.upper()}")
print(f"Malá písmena: {second.lower()}")
print(f"Prvních 5 znaků: {second[:5]}")
```

### Boolean (bool) - pravdivostní hodnoty

```python
# Boolean (True/False)
is_true = True
is_false = False

print(f"is_true = {is_true}, typ: {type(is_true)}")
print(f"is_false = {is_false}")

# Porovnání
print(f"\n5 > 3: {5 > 3}")
print(f"10 == 10: {10 == 10}")
print(f"7 != 8: {7 != 8}")
```

```python
# Logické operátory
x = True
y = False

print(f"x AND y: {x and y}")
print(f"x OR y: {x or y}")
print(f"NOT x: {not x}")
print(f"NOT y: {not y}")
```

### None - prázdná hodnota

```python
# None - speciální hodnota pro "nic"
nothing = None

print(f"Hodnota: {nothing}, typ: {type(nothing)}")
print(f"Je None? {nothing is None}")
print(f"Není None? {nothing is not None}")
```

## 2. Kolekce (Collections)

Python má čtyři základní typy kolekcí.

### List - měnitelný seznam

```python
# List - seřazená měnitelná kolekce
fruits = ["jablko", "hruška", "banán"]
numbers = [1, 2, 3, 4, 5]
mixed = [1, "text", 3.14, True, None]  # může obsahovat různé typy

print(f"Ovoce: {fruits}")
print(f"Typ: {type(fruits)}")
print(f"Délka: {len(fruits)}")
print(f"První prvek: {fruits[0]}")
print(f"Poslední prvek: {fruits[-1]}")
```

```python
# Operace s listy
numbers = [1, 2, 3]

# Přidání
numbers.append(4)
print(f"Po append(4): {numbers}")

# Vložení na pozici
numbers.insert(0, 0)
print(f"Po insert(0, 0): {numbers}")

# Odebrání
numbers.remove(2)
print(f"Po remove(2): {numbers}")

# Pop - odebere a vrátí poslední prvek
last = numbers.pop()
print(f"Pop vrátil: {last}, zůstalo: {numbers}")
```

### Tuple - neměnitelná n-tice

```python
# Tuple - seřazená neměnitelná kolekce
coordinates = (10, 20)
rgb = (255, 128, 0)
single = (42,)  # čárka je důležitá pro jeden prvek!

print(f"Souřadnice: {coordinates}, typ: {type(coordinates)}")
print(f"RGB: {rgb}")
print(f"X: {coordinates[0]}, Y: {coordinates[1]}")
```

```python
# Tuple unpacking - rozbalení
point = (100, 200)
x, y = point
print(f"x = {x}, y = {y}")

# Výměna hodnot pomocí tuple
a = 5
b = 10
print(f"Před výměnou: a={a}, b={b}")
a, b = b, a
print(f"Po výměně: a={a}, b={b}")
```

### Dictionary (dict) - slovník (klíč-hodnota)

```python
# Dictionary - neseřazená kolekce párů klíč: hodnota
person = {
    "jméno": "Jan",
    "věk": 30,
    "město": "Praha"
}

print(f"Osoba: {person}")
print(f"Typ: {type(person)}")
print(f"Jméno: {person['jméno']}")
print(f"Věk: {person['věk']}")
```

```python
# Operace se slovníky
data = {"a": 1, "b": 2}

# Přidání/změna
data["c"] = 3
print(f"Po přidání 'c': {data}")

# Získání s defaultní hodnotou
print(f"Klíč 'a': {data.get('a')}")
print(f"Klíč 'x' (neexistuje): {data.get('x', 'nenalezeno')}")

# Klíče, hodnoty, páry
print(f"Klíče: {list(data.keys())}")
print(f"Hodnoty: {list(data.values())}")
print(f"Páry: {list(data.items())}")
```

### Set - množina (unikátní prvky)

```python
# Set - neseřazená kolekce unikátních prvků
numbers = {1, 2, 3, 4, 5}
duplicates = {1, 2, 2, 3, 3, 3}  # automaticky odstraní duplicity

print(f"Čísla: {numbers}, typ: {type(numbers)}")
print(f"S duplicity: {duplicates}")  # bude {1, 2, 3}
print(f"Je 3 v množině? {3 in numbers}")
```

```python
# Množinové operace
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(f"A: {a}")
print(f"B: {b}")
print(f"Sjednocení (A | B): {a | b}")
print(f"Průnik (A & B): {a & b}")
print(f"Rozdíl (A - B): {a - b}")
print(f"Symetrický rozdíl (A ^ B): {a ^ b}")
```

## 3. Type Checking a konverze

Zjišťování typů a převody mezi nimi.

```python
# Zjištění typu pomocí type()
x = 42
y = "hello"
z = [1, 2, 3]

print(f"type(42): {type(x)}")
print(f"type('hello'): {type(y)}")
print(f"type([1,2,3]): {type(z)}")

# Kontrola typu pomocí isinstance()
print(f"\nJe x integer? {isinstance(x, int)}")
print(f"Je y string? {isinstance(y, str)}")
print(f"Je z list? {isinstance(z, list)}")
```

```python
# Konverze mezi typy
text_number = "123"
text_float = "45.67"

# String -> int/float
num = int(text_number)
decimal = float(text_float)
print(f"String '{text_number}' -> int: {num}")
print(f"String '{text_float}' -> float: {decimal}")

# int/float -> string
number = 42
text = str(number)
print(f"int {number} -> string: '{text}'")

# float -> int (zaokrouhlí dolů)
pi = 3.14159
print(f"float {pi} -> int: {int(pi)}")
```

```python
# Konverze kolekcí
my_list = [1, 2, 2, 3, 3, 3]
my_tuple = (4, 5, 6)
my_string = "hello"

# List -> Set (odstraní duplicity)
unique = set(my_list)
print(f"List {my_list} -> Set: {unique}")

# Tuple -> List
as_list = list(my_tuple)
print(f"Tuple {my_tuple} -> List: {as_list}")

# String -> List (jednotlivé znaky)
chars = list(my_string)
print(f"String '{my_string}' -> List: {chars}")

# List -> String
joined = ''.join(chars)
print(f"List {chars} -> String: '{joined}'")
```

## 4. Mutable vs Immutable

Důležitý koncept - některé typy lze měnit, jiné ne.

```python
# IMMUTABLE (neměnitelné) - int, float, str, tuple, bool, None
x = 10
print(f"Původní x: {x}, id: {id(x)}")

x = x + 1  # vytvoří se NOVÝ objekt!
print(f"Nové x: {x}, id: {id(x)}")
print("Všimni si změny ID - je to nový objekt!\n")

# String je také immutable
text = "hello"
print(f"Původní text: {text}, id: {id(text)}")
text = text.upper()  # vytvoří nový string
print(f"Nový text: {text}, id: {id(text)}")
```

```python
# MUTABLE (měnitelné) - list, dict, set
my_list = [1, 2, 3]
print(f"Původní list: {my_list}, id: {id(my_list)}")

my_list.append(4)  # mění STEJNÝ objekt!
print(f"Po append: {my_list}, id: {id(my_list)}")
print("ID je stejné - je to stále ten samý objekt!\n")

# Pozor na reference!
list1 = [1, 2, 3]
list2 = list1  # list2 odkazuje na STEJNÝ objekt
print(f"list1: {list1}")
print(f"list2: {list2}")

list2.append(4)
print(f"\nPo list2.append(4):")
print(f"list1: {list1}")  # list1 se také změnil!
print(f"list2: {list2}")
```

```python
# Vytvoření kopie (pro mutable typy)
original = [1, 2, 3]
copy = original.copy()  # nebo original[:]

print(f"Original: {original}, id: {id(original)}")
print(f"Copy: {copy}, id: {id(copy)}")
print(f"Jsou různé objekty? {id(original) != id(copy)}\n")

copy.append(4)
print(f"Po copy.append(4):")
print(f"Original: {original}")  # zůstal nezměněný
print(f"Copy: {copy}")  # změnil se jen copy
```

## 5. Praktické příklady a tipy

```python
# List comprehension - elegantní tvorba listů
numbers = [1, 2, 3, 4, 5]

# Klasický způsob
squares = []
for n in numbers:
    squares.append(n ** 2)
print(f"Klasicky: {squares}")

# List comprehension
squares = [n ** 2 for n in numbers]
print(f"List comprehension: {squares}")

# S podmínkou - jen sudá čísla
evens = [n for n in numbers if n % 2 == 0]
print(f"Sudá čísla: {evens}")
```

```python
# Dictionary comprehension
numbers = [1, 2, 3, 4, 5]

# Vytvoření slovníku číslo: druhá mocnina
squares_dict = {n: n**2 for n in numbers}
print(f"Slovník mocnin: {squares_dict}")

# Set comprehension
unique_squares = {n**2 for n in [1, 2, 2, 3, 3]}
print(f"Unikátní mocniny: {unique_squares}")
```

```python
# Unpacking a *args
data = [1, 2, 3, 4, 5]

# Rozbalení prvních prvků
first, second, *rest = data
print(f"První: {first}")
print(f"Druhý: {second}")
print(f"Zbytek: {rest}")

# První a poslední
first, *middle, last = data
print(f"\nPrvní: {first}")
print(f"Prostřední: {middle}")
print(f"Poslední: {last}")
```

```python
# Užitečné operace s kolekcemi
numbers = [5, 2, 8, 1, 9, 3]

print(f"List: {numbers}")
print(f"Minimum: {min(numbers)}")
print(f"Maximum: {max(numbers)}")
print(f"Součet: {sum(numbers)}")
print(f"Délka: {len(numbers)}")
print(f"Seřazený: {sorted(numbers)}")
print(f"Obráceně: {sorted(numbers, reverse=True)}")
```

## Shrnutí

### Základní typy
| Typ | Popis | Příklad | Mutable |
|-----|-------|---------|----------|
| `int` | Celá čísla | `42` | Ne |
| `float` | Desetinná čísla | `3.14` | Ne |
| `str` | Textové řetězce | `"hello"` | Ne |
| `bool` | True/False | `True` | Ne |
| `None` | Prázdná hodnota | `None` | Ne |

### Kolekce
| Typ | Popis | Příklad | Mutable | Seřazené |
|-----|-------|---------|---------|----------|
| `list` | Seznam prvků | `[1, 2, 3]` | Ano | Ano |
| `tuple` | Neměnitelná n-tice | `(1, 2, 3)` | Ne | Ano |
| `dict` | Slovník klíč-hodnota | `{"a": 1}` | Ano | Ne (Python 3.7+: zachovává pořadí vložení) |
| `set` | Množina (unikátní) | `{1, 2, 3}` | Ano | Ne |

### Kdy použít co?
- **List**: Když potřebuješ seřazenou kolekci, kterou budeš měnit
- **Tuple**: Pro data, která se nemění (souřadnice, RGB barvy)
- **Dict**: Pro mapování klíčů na hodnoty (konfigurace, databáze)
- **Set**: Pro unikátní prvky a množinové operace
